```python
#1 Print all prime numbers between an input range
start, end = map(int, input().split())
for num in range(start, end + 1):
    if num > 1:
        for i in range(2, int(num ** 0.5) + 1):
            if num % i == 0:
                break
        else:
            print(num, end=" ")
```

```python
# 2 Factorial using recursion
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

n = int(input("Enter a number: "))

if n < 0:
    print("Factorial is not defined")
else:
    print("Factorial:", factorial(n))
```

```python
#3. Square of numbers using lambda
square = lambda x: x * x
numbers = list(map(int, input().split()))
result = list(map(square, numbers))
print("Squares:", result)
```

```python
# 4. Find the second largest element in a list
numbers = list(map(int, input().split()))
unique = list(set(numbers))
unique.sort(reverse=True)
if len(unique) >= 2:
    print("Second Largest:", unique[1])
else:
    print("No second largest element")
```

```python
# Count the frequency of characters in a string
s = input("Enter a string: ")
freq = {}
for ch in s:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
for ch, count in freq.items():
    print(ch, ":", count)
```

```python
# Calculate the area of a circle using the math library
import math
r = float(input("Enter radius: "))
area = math.pi * r * r
print("Area:", round(area, 2))
```

```python
# Reverse a string without using built-in reverse
s = input("Enter a string: ")
rev = ""
for ch in s:
    rev = ch + rev
print("Reversed String:", rev)
```

```python
# Remove duplicates from a list
numbers = list(map(int, input().split()))
result = []
for num in numbers:
    if num not in result:
        result.append(num)
print("After Removing Duplicates:", result)
```

```python
# Merge two dictionaries
dict1 = {"a": 10, "b": 20}
dict2 = {"c": 30, "d": 40}
merged = dict1.copy()
merged.update(dict2)
print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged Dictionary:", merged)
```

```python
# Fibonacci series using recursion
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
n = int(input("Enter number of terms: "))
if n <= 0:
    print("Enter a positive number")
else:
    for i in range(n):
        print(fibonacci(i), end=" ")
```
